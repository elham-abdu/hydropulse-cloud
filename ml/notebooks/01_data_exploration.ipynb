```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Load data
df = pd.read_csv('L-TOWN_pressures.csv')

# 2. Inspect shape & first few rows
print("Dataset Shape:", df.shape)
print("\nFirst 5 Rows:")
print(df.head())

# 3. View summary statistics
print("\nSummary Statistics:")
print(df.describe())

# 4. Plot sensor readings sample
plt.figure(figsize=(12, 4))
df.iloc[:, 1].head(500).plot(title="L-TOWN Sensor Pressure Readings")
plt.xlabel("Reading Index")
plt.ylabel("Pressure")
plt.grid(True)
plt.show()
```
